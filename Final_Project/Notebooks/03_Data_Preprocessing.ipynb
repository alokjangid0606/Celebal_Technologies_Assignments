```python
import pandas as pd
import numpy as np
import json
import re
```

## Load Dataset

```python
data = []

with open("../data/resumes_dataset.jsonl", "r", encoding="utf-8") as file:
    for line in file:
        data.append(json.loads(line))

df = pd.DataFrame(data)

print(df.shape)
```

## Check One Resume

```python
print(df.loc[0, "Text"][:1000])
```

## Create Cleaning Function

```python
def clean_resume(text):

    text = str(text)

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove Email
    text = re.sub(r"\S+@\S+", "", text)

    # Remove Phone Number
    text = re.sub(r"\+?\d[\d\s\-\(\)]{8,}", "", text)

    # Remove Special Characters
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)

    # Remove Multiple Spaces
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

## Apply Cleaning

```python
# Apply Cleaning Function

df["Clean_Text"] = df["Text"].apply(lambda x: clean_resume(x))

print("Clean_Text column created successfully.")

df[["Text", "Clean_Text"]].head()
```

## Compare Original vs Clean Text

```python
print("="*80)
print("ORIGINAL TEXT\n")
print(df.loc[0, "Text"][:700])

print("\n" + "="*80)

print("CLEAN TEXT\n")
print(df.loc[0, "Clean_Text"][:700])
```

## Text Length Comparison

```python
df["Original_Length"] = df["Text"].apply(len)
df["Clean_Length"] = df["Clean_Text"].apply(len)

df[["Original_Length", "Clean_Length"]].head()
```

## Check if Any Empty Resume Exists

```python
print("Empty Clean Resumes :", (df["Clean_Text"].str.strip() == "").sum())
```

## Check Duplicate Resumes After Cleaning

```python
duplicates = df["Clean_Text"].duplicated().sum()

print("Duplicate Clean Resumes :", duplicates)
```

## Select Important Columns

```python
final_df = df[
    [
        "ResumeID",
        "Category",
        "Skills",
        "Education",
        "Experience",
        "Clean_Text",
        "Text",
        "Source"
    ]
]

final_df.head()
```

## Final Dataset Information

```python
print(final_df.shape)

final_df.info()
```

## Save Clean Dataset

```python
final_df.to_csv("../data/final_resume_dataset.csv", index=False)

print("Final Resume Dataset Saved Successfully")
```
