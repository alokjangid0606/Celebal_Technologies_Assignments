```python
import pandas as pd
```

## Load ATS Report

```python
ats = pd.read_csv("../outputs/ats_report.csv")

ats
```

## Load Skill Gap Report

```python
skill = pd.read_csv("../outputs/skill_gap_report.csv")

skill
```

### Extract Values

```python
ats_score = ats.loc[0, "Final ATS"]

matched_skills = skill.loc[0, "Matched Skills"]

missing_skills = skill.loc[0, "Missing Skills"]

skill_percentage = skill.loc[0, "Skill Match %"]
```

### Strength Function

```python
strengths = []

if ats_score >= 80:
    strengths.append("Good ATS Score")

if skill_percentage >= 70:
    strengths.append("Good Skill Match")

if matched_skills != "":
    strengths.append("Relevant Technical Skills Found")
```

### Weakness Function

```python
weaknesses = []

if ats_score < 70:
    weaknesses.append("ATS Score is Low")

if skill_percentage < 60:
    weaknesses.append("Important Skills are Missing")

if missing_skills != "":
    weaknesses.append("Resume Missing Required Skills")
```

### Recommendation Function

```python
recommendations = []

if "docker" in missing_skills.lower():
    recommendations.append("Learn Docker")

if "aws" in missing_skills.lower():
    recommendations.append("Learn AWS")

if "machine learning" in missing_skills.lower():
    recommendations.append("Improve Machine Learning Skills")

if ats_score < 70:
    recommendations.append("Customize Resume According to Job Description")
```

### Print AI Feedback

```python
print("="*60)

print("AI RESUME FEEDBACK")

print("="*60)

print()

print("Overall ATS Score :", ats_score)

print()

print("Strengths")

for s in strengths:
    print(">", s)

print()

print("Weaknesses")

for w in weaknesses:
    print(">", w)

print()

print("Recommendations")

for r in recommendations:
    print("•", r)

print()

print("="*60)
```

### Create Final Report

```python
report = {

    "ATS Score": ats_score,

    "Strengths": ", ".join(strengths),

    "Weaknesses": ", ".join(weaknesses),

    "Recommendations": ", ".join(recommendations)

}

pd.DataFrame([report])
```

```python
pd.DataFrame([report]).to_csv(
    "../outputs/ai_feedback_report.csv",
    index=False
)

print("AI Feedback Report Saved")
```
