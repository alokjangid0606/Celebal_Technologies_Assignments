## Import Libraries

```python
import pandas as pd
import re
```

## Load Dataset

```python
df = pd.read_csv("../data/final_feature_dataset.csv")

print(df.shape)

df.head()
```

## Select Resume

```python
resume = df.loc[0, "Clean_Text"]

print(resume[:500])
```

## Job Description

```python
job_description = """
We are looking for a Python Developer.

Required Skills:

Python
Machine Learning
SQL
Git
Docker
AWS
REST API
Linux

Good Communication Skills
"""
```

## Master Skill List

```python
MASTER_SKILLS = [

    "python","java","c","c++","sql",

    "machine learning","deep learning",

    "tensorflow","pytorch",

    "numpy","pandas",

    "scikit-learn",

    "docker","kubernetes",

    "aws","azure","gcp",

    "git","github",

    "linux",

    "mongodb","mysql","postgresql",

    "html","css","javascript",

    "react","nodejs",

    "flask","django",

    "power bi","tableau",

    "excel",

    "rest api"

]
```

## Skill Extraction Function

```python
def extract_skills(text):

    text = text.lower()

    found = []

    for skill in MASTER_SKILLS:

        if re.search(r"\b" + re.escape(skill) + r"\b", text):

            found.append(skill)

    return sorted(list(set(found)))
```

## Extract Skills

```python
resume_skills = extract_skills(resume)

jd_skills = extract_skills(job_description)

print("Resume Skills")

print(resume_skills)

print()

print("JD Skills")

print(jd_skills)
```

## Matched & Missing Skills

```python
matched_skills = list(
    set(resume_skills).intersection(jd_skills)
)

missing_skills = list(
    set(jd_skills).difference(resume_skills)
)

print("Matched Skills")

print(matched_skills)

print()

print("Missing Skills")

print(missing_skills)
```

## Skill Match Percentage

```python
skill_percentage = (
    len(matched_skills) /
    len(jd_skills)
) * 100

print(f"Skill Match : {skill_percentage:.2f}%")
```

## Skill Report

```python
report = {

    "Resume Skills": resume_skills,

    "JD Skills": jd_skills,

    "Matched Skills": matched_skills,

    "Missing Skills": missing_skills,

    "Skill Match %": round(skill_percentage,2)

}

report
```

## Save Report

```python
report_df = pd.DataFrame({

    "Matched Skills":[", ".join(matched_skills)],

    "Missing Skills":[", ".join(missing_skills)],

    "Skill Match %":[round(skill_percentage,2)]

})

report_df.to_csv(
    "../outputs/skill_gap_report.csv",
    index=False
)

print("Skill Gap Report Saved")
```
