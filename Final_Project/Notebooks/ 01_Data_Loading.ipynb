# Import Libraries

```python
import json
import pandas as pd
import numpy as np

print("Libraries Imported Successfully")
```

## Read JSONL Dataset

```python
data = []

with open("../data/resumes_dataset.jsonl", "r", encoding="utf-8") as file:
    for line in file:
        data.append(json.loads(line))

print("Total Resumes :", len(data))
```

## Convert into DataFrame

```python
df = pd.DataFrame(data)

df.head()
```
